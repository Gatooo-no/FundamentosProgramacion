```python
Tabla = int(input("¿Que tabla de multipicar quieres? "))
Inicio = int(input("¿Por donde empieza? "))
Fin = int(input("¿Hasta donde acaba? "))
#While es un bucle hasta que se cumpla la peticion como debe de ser
while Inicio >= Fin:
  print("El limite final debe de ser mayor que el inicial")
  Fin = int(input("¿Hasta donde acaba? "))
Incremenetos = int(input("¿De cuanto en cuanto? "))
for i in range(Inicio, Fin+1, Incremenetos):
  print(Tabla, "*", i, "=", Tabla*i)
```

#**Tema Listas en Pyton**

```python
alumnos = ["Ana", "Luis", "Carlos"]
print(alumnos)
print(alumnos[2])
print(alumnos[0])
```

```python
print(alumnos)
print(alumnos[2])
print(alumnos[1])
```

```python
#append pone un nuevo elento en la lista
print(alumnos)
alumnos.append("Sandra")
print(alumnos)
```

```python

print(alumnos)
alumnos.insert (2, "Sofia")
print(alumnos)
```

```python
print (alumnos)
alumnos.remove("Sofia")
print (alumnos)
```

```python
print (alumnos [2])
print (alumnos [7])
#Como solo hay 5 elementos no existe el alumno 7
```

```python
#.pop borra a un elemento especifico de la lista
print(alumnos)
alumnos.pop(2)
print(alumnos)
```

```python
print(alumnos)
del alumnos[2]
print(alumnos)
#del tambien elemina elementos
```

```python
#len cuenta los artculos de la lista
print("Hay un total de:", len(alumnos), "alumnos en la lista")
```

```python
#for es un ciclo para recorer la lista  y hacer un proceso
#break hace que se termine de leer la lista
for Y in alumnos:
  print(Y)
  if Y == "Sofia":
    print("^^ MIRA ESTA ES SOFIA")
    break
```
