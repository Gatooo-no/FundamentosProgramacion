```python
def tablas_de_multilicar(x):
  for i in range(11):
    print(i, "por", x, "es", i*x)
tablas_de_multilicar(3)
tablas_de_multilicar(5)
```

```python
#Definir una funcion que calcule el area de un triangulo
#La funcion recive dos parametros, La Base Y La Altura
def calcular_area(base, altura):
  area = base * altura
  return area

resultado = calcular_area(26, 14)
print("Area del triangulo: ", resultado)
print("Area del triangulo: ", calcular_area(10, 5))
```

```python
def miFuncion():
#La vareablex de miFuncion existe unicamente dentro del
#Codigo de la funcion
  variablex = 123
  print(variablex)

miFuncion()
#La siguiente variable solo existe en miFuncion.
print(variablex)
```

```python
def miFuncion():
#La vareablex de miFuncion existe unicamente dentro del
#Codigo de la funcion
  global variablex
  variablex = 123
  print(variablex)

miFuncion()
#La siguiente variable solo existe en miFuncion.
print(variablex)
```
